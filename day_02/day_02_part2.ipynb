We will use the **nf-core/rnaseq** pipeline to analyse the two downloaded RNA-seq samples.

**What does the pipeline do?**

The pipeline processes raw FASTQ files and performs the main steps of an RNA-seq analysis:

- quality control of the raw sequencing reads,
- adapter and quality trimming,
- alignment of reads to a reference genome,
- quantification of gene/transcript expression,
- additional quality-control checks,
- generation of a gene expression/count matrix,
- and collection of the QC results in a MultiQC report.

**Main tools**

Some of the main tools used by the pipeline are:

- **FastQC** – checks the quality of the raw sequencing reads.
- **Trim Galore!** – removes adapters and low-quality sequence.
- **STAR** – aligns RNA-seq reads to the reference genome.
- **Salmon** – quantifies transcript/gene expression.
- **SAMtools** – processes and evaluates alignment files.
- **RSeQC / Qualimap** – perform additional quality-control checks on the aligned reads.
- **MultiQC** – combines the QC results from the different tools into one report.

*Explain all the parameters you set and why you set them in this way.*



- `-r 3.27.0`
  - Specifies the exact version of the nf-core/rnaseq pipeline.
  - Using a fixed version improves reproducibility because future versions of the pipeline may behave differently.

- `-profile docker`
  - Runs the required software in Docker containers.
  - This provides predefined software environments and versions and reduces differences between computer systems.

- `-c rnaseq.config`
  - Loads our custom Nextflow configuration file.
  - We use it to limit the computational resources available to individual processes.
  - In our configuration, processes are limited to a maximum of 4 CPUs and 10 GB of memory.

- `-work-dir ~/nextflow_work`
  - Specifies the directory used by Nextflow for temporary and intermediate workflow files.
  - The work directory was moved to the Linux filesystem because STAR requires FIFO files, which are not supported correctly in the Windows-mounted NTFS filesystem used through WSL.

- `--input rnaseq_samplesheet.csv`
  - Specifies the samplesheet containing our two samples.
  - The samplesheet connects the sample names to their paired-end FASTQ files.

- `--outdir rnaseq_results`
  - Specifies the directory in which the pipeline results will be stored.

- `--fasta reference_chr19/Mus_musculus.GRCm38.dna.chromosome.19.fa.gz`
  - Specifies the FASTA file containing the reference DNA sequence.
  - We use only chromosome 19 of the GRCm38 mouse genome as a reduced reference.
  - This reduces the computational resources required for the analysis.

- `--gtf reference_chr19/Mus_musculus.GRCm38.102.chr19.gtf.gz`
  - Specifies the GTF file containing the gene annotations for the reference.
  - The original GRCm38 annotation was filtered to contain only chromosome 19.
  - The FASTA and GTF therefore contain the same chromosome.

- `--aligner star_salmon`
  - Uses STAR for alignment and Salmon for RNA-seq quantification.
  - STAR aligns the reads against the chromosome 19 reference.
  - Using the reduced chromosome 19 reference makes the STAR workflow less computationally demanding than using the complete mouse genome.

- `-resume`
  - Allows Nextflow to reuse successfully completed processes from a previous run.
  - If the pipeline is interrupted or fails, already completed steps do not have to be repeated.

**Quality control**

FastQC was performed before and after trimming to assess sequence quality, adapter contamination, GC content and sequence duplication. The raw reads showed high per-base and per-sequence quality, but adapter contamination was detected in all four FASTQ files. After processing with Trim Galore, adapter content passed for all four reads, indicating successful adapter removal.

*rRNA contamination:*  
High rRNA contamination is undesirable because reads originating from rRNA reduce the effective sequencing depth available for informative mRNA transcripts. The available FastQC results do not directly quantify rRNA contamination, so no conclusion about its level in these samples can be made.

*Sequence duplication:*  
Duplication differed considerably between samples. After theoretical deduplication, 68.69% and 75.02% of reads remained for Sham-Oxy, compared with only 38.10% and 45.27% for SNI-Oxy. The higher duplication in SNI-Oxy may indicate low library complexity or PCR amplification bias. However, some duplication is expected in RNA-seq due to highly expressed transcripts, so FastQC alone cannot determine whether this is a technical problem.

*GC content:*  
All four trimmed reads received a GC-content warning. The main distribution was centered around 45–50% GC, but an additional GC-rich population around 80–87% was visible. This may reflect RNA-seq library composition, but technical bias or contamination cannot be excluded.

Overall, the data appear suitable for further analysis: sequence quality is high and adapter contamination was successfully removed. However, the high duplication in SNI-Oxy and the unusual GC-content distributions should be considered when interpreting downstream results.

**Possible causes of poorer results**

Poorer RNA-seq results can arise at several stages of the experiment. RNA degradation or poor sample quality can affect the original material, while inefficient rRNA depletion or mRNA enrichment can reduce the amount of informative RNA. During library preparation, excessive PCR amplification can increase duplication and adapter contamination may be introduced. Sequencing can additionally produce low-quality reads, while inappropriate trimming, contamination, or an incorrect reference genome or annotation can negatively affect downstream analysis.
